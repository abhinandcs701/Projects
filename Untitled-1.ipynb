```python
#STUDENT PERFOMANCE REPORT AND ANALYSIS
''' ##Overview:- 
This project analyzes a dataset of 1,000 students' exam scores (math, reading, 
writing) alongside  background factors (gender, parental education, 
lunch type, test preparation) to understand what impacts academic performance.

## Questions Explored
- Do background factors (parental education, lunch type, test prep) have a 
  statistically significant and practically meaningful effect on scores?
- Can a student's math score be predicted from their reading/writing performance?
- Do students naturally cluster into distinct performance tiers, and do those 
  tiers relate to background factors?

## Methods
Exploratory data analysis, hypothesis testing (t-tests, ANOVA, chi-square), 
effect size analysis (Cohen's d), linear regression, logistic regression, 
decision trees, and K-Means clustering.

## Key Findings
- Parental education has the largest practical effect on scores (Found out by Cohen's d = 0.78), 
  followed by lunch type (0.63) and test prep completion (0.55)
- Test preparation completion is the only factor students can directly control, 
  and it still produces a meaningful, statistically significant score boost.
- A simple 3-feature regression model (reading, writing, test prep) performs 
  nearly as well as a full 9-feature model, suggesting most predictive power 
  comes from a few core variables.
- Unsupervised clustering, without ever seeing background data, independently 
  discovered performance tiers that align with known socioeconomic factors 
  (lunch type).  '''


import numpy as np  
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sea 
import scipy.stats as stats
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree



#DATA INSPECTION:

k=pd.read_csv('studentperfomance.csv')
print(k.head(5))
print('columns:  ',k.columns)
print('shape:  ',k.shape)
#print('INfo',k.info())

'''##CLENSING DATA 
    no duplicate rows or NULL values were found in this dataset and every column is according to its own data type '''

print(' Checking for NULL values and viewing dtype  ')
print(k.isnull().sum())
k.duplicated().sum()
k.dtypes
k['total score']=(k['math score']+k['reading score']+k['writing score'])

# Average scores by parental education level,
print(k.groupby('parental level of education')['total score'].mean().sort_values(ascending=False))
# Test prepration completion effect, by lunch type,
print(k.groupby(['lunch','test preparation course'])['total score'].mean().unstack())

'''Parental education shows a clear gradient:
Students with a parent holding a 
-master's degree average  aprox 31 points higher total score than those whose 
  parent's highest education is high school.
- Completing the test preparation course is associated with a 21-26 point 
  score increase, regardless of the  lunch type.
- Only a few  of students (23 of 1000) scored above 90 in all three subjects.'''


'''sea.histplot(k['total score'],bins=20,kde=True)   #Total score histogram 
plt.show()


sea.barplot(data=k,x='parental level of education', y='total score')  #mean Barplot
plt.xticks(rotation=45)
plt.title('Barplot of parental education')
plt.show()

sea.boxplot(data=k,x='lunch',y='total score',hue='test preparation course')
plt.title(' Boxplot -Total score vs lunch ')
plt.show()



corr_box = k[['math score','reading score','writing score']].corr()
sea.heatmap(corr_box,annot=True,cmap='coolwarm')    #HEATMAP    ///
plt.title('Correlation Between Subjects')
plt.show()

sea.scatterplot(data=k,x='math score',y='reading score',hue='gender')
plt.title('MATH VS READING SCATTERPLOT HUE BY GENDER')
plt.show()'''

#tests

#T-test
male_score=k[k['gender']=='male']['math score']
female_score=k[k['gender']=='female']['math score']
t_stats,p_value=stats.ttest_ind(male_score,female_score)
print('t_stats',t_stats,'p_value',p_value)

'''##MALE SCORERS are significantly higher on average (t-stat=0.538,p-val,0.001), signifing that the diffrence is real and indeed not due to
  random chance.'''

#ANOVA test
race_groups=[k[k['race/ethnicity']==i]['total score'] for i in k['race/ethnicity'].unique()]
t_stat,p_val=stats.f_oneway(*race_groups)
print('t_stat.race',t_stat,'p_value.race',p_val)

groups_edu = [k[k['parental level of education']==i]['total score'] for i in k['parental level of education'].unique()]
f_edu, p_edu = stats.f_oneway(*groups_edu)
print('t_stat.edu',f_edu,'p_val.edu',p_edu)

'''**Finding:** both race/ethnicity (F=9.10, p<0.0001) and parental education 
(F=10.75, p<0.0001) show statistically significant differences in total score 
across groups and at least one group in each category performs differently from 
the rest.'''

#CHI square test 
con_table=pd.crosstab(k['lunch'],k['test preparation course'])
chi2,p_chi,dof,expected=stats.chi2_contingency(con_table)
print('pchi',p_chi,'chi2',chi2)

#Coherens D
print('chorens d ')
def cohren(a,b):
    a1,b1=len(a),len(b)
    pool=(((a1-1)*a.var())+((b1-1)*b.var()))/(a1+b1-2)
    pool=pool**0.5
    return (a.mean()-b.mean())/pool

d_parental = cohren(
    k[k['parental level of education']=="master's degree"]['total score'],
    k[k['parental level of education']=='high school']['total score']
)
d_lunch = cohren(
    k[k['lunch']=='standard']['total score'],
    k[k['lunch']=='free/reduced']['total score']
)
d_prep = cohren(
    k[k['test preparation course']=='completed']['total score'],
    k[k['test preparation course']=='none']['total score']
)

print("Parental education (master's vs high school)",d_parental)
print('Lunch type (standard vs free/reduced)',d_lunch)
print('Test prep (completed vs none)',d_prep)


#BASELINE MODEL : 2-features
x_base=k[['reading score','writing score']]
y_ans=k['math score']
x_train,x_test,y_train,y_test=train_test_split(x_base,y_ans,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(x_train,y_train)
predict=model.predict(x_test)
mae=mean_absolute_error(y_test,predict)
rmse=mean_squared_error(y_test,predict)**0.5
r2=r2_score(y_test,predict)
print('LINEAR model 1','mae',mae,'rmse',rmse,'r2',r2)


#9feature model

k_encode=pd.get_dummies(k,columns=['parental level of education','lunch'],drop_first=True)
k_encode['test_mapp']=k['test preparation course'].map({'none':0,'completed':1})
feature_cols=['writing score','reading score','test_mapp']+[col for col in k_encode.columns if 'parental level of education_' in col or 'lunch_' in col]
x1=k_encode[feature_cols]
y1=k_encode['math score']
x1_train,x1_test,y1_train,y1_test=train_test_split(x_base,y_ans,test_size=0.2,random_state=42)
model1=LinearRegression()
model1.fit(x1_train,y1_train)
predict1=model1.predict(x1_test)
mae=mean_absolute_error(y1_test,predict1)
rmse=mean_squared_error(y1_test,predict1)**0.5
r2=r2_score(y1_test,predict1)
print('9 Feature model ','mae',mae,'rmse',rmse,'r2',r2)

#3Feature model 
x2=k_encode[['test_mapp','writing score','reading score']]
y2=k_encode['math score']
x2_train,x2_test,y2_train,y2_test=train_test_split(x_base,y_ans,test_size=0.2,random_state=42)
model2=LinearRegression()
model2.fit(x2_train,y2_train)
predict2=model2.predict(x2_test)
mae=mean_absolute_error(y2_test,predict2)
rmse=mean_squared_error(y2_test,predict2)**0.5
r2=r2_score(y2_test,predict2)
print('3-Simplified  Feature model ','mae',mae,'rmse',rmse,'r2',r2)


#LOGICAL MODEL
print('LOGICAL model')

k['pass']=(k['total score']>=180).astype(int)
k['pass'].value_counts(normalize=True)
x_clf=k[['writing score','reading score']]
y_clf=k['pass']
X_train,X_test,Y_train,Y_test=train_test_split(x_clf,y_clf,random_state=42,test_size=0.2)
clf_model=LogisticRegression()
clf_model.fit(X_train,Y_train)
clf_pred=clf_model.predict(X_test)
print('accuracy',accuracy_score(Y_test,clf_pred))
print(classification_report(Y_test,clf_pred))



#decition tree
tree_clf = DecisionTreeClassifier(max_depth=4, random_state=42)
tree_clf.fit(X_train, Y_train)
tree_preds = tree_clf.predict(X_test)
print('Decision Tree accuracy:', accuracy_score(Y_test, tree_preds))

'''plt.figure(figsize=(18, 8))
plot_tree(tree_clf, feature_names=['reading score','writing score'], 
          class_names=['fail','pass'], filled=True, fontsize=10)
plt.title('Decision Tree: Pass/Fail Prediction')
plt.show()'''


#Unsupervised learning
x_cluster= k[['math score','reading score','writing score']]
scale=StandardScaler()
x_scaled=scale.fit_transform(x_cluster)
Kmean=KMeans(n_clusters=3,random_state=42)
k['cluster']=Kmean.fit_predict(x_scaled)
print(k['cluster'].value_counts())
print(k.groupby(by='cluster')[['math score','reading score','writing score']].mean())
print(pd.crosstab(k['cluster'], k['lunch'], normalize='index'))

'''##K-Means is distance-based, so unscaled features could bias the clustering if one column 
happened to have a larger numeric range — scaling puts all three subjects on equal footing.'''
'''##The lowest-performing cluster has 55.8% free/reduced lunch students, vs. 23.1% in the highest-performing cluster — even though lunch was never used to create the clusters.
This is the standout finding of this section: two entirely different methods (Week 3's Cohen's d, and Week 5's unsupervised clustering).
independently arrived at the same conclusion about lunch type's relationship to performance.'''

''''##  Conclusions

- **Background factors matter, and can be ranked by impact:** parental education 
  (d=0.78) > lunch type (d=0.63) > test prep completion (d=0.55). All three are 
  statistically significant.
- **Test prep is the one actionable factor:** unlike parental education or lunch 
  type, it's something a student can directly influence, and it produces a real, 
  measurable score improvement.
- **Simpler models often perform nearly as well as complex ones:** a 3-feature 
  regression model came within 0.01 R² of a 9-feature model, since background factors 
  were already partially captured by reading/writing scores.
- **Two independent methods (statistical testing and unsupervised clustering) 
  converged on the same conclusion** regarding the relationship between lunch type 
  and performance — a strong form of validation.
- **Logistic Regression outperformed Decision Trees** for this specific pass/fail 
  prediction task, likely due to the near-linear relationship between features and 
  outcome; deeper trees overfit without improving generalization.

## Limitations
- This dataset does not indicate the source, region, or time period of collection, 
  limiting how broadly these findings can be generalized.
- Statistical significance does not imply causation — background factors are 
  associated with performance, not proven to cause it.
- The pass/fail threshold (180/300) was arbitrarily chosen for demonstration purposes.'''
```
